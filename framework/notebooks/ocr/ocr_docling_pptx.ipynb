### CONVERT PPT/PPTX

### Qwen3-VL-8B-Instruct

WITHOUT IMAGE DESCRIPTION

```python
docling_converter = DoclingVLMConverter(
    model= "Qwen3-VL-8B-Instruct",
    hostname_and_port="https://llmservice.air.id/chat/completions",
    prompt = "Convert this page to markdown.",
    api_key="",
    response_format = "markdown",
    temperature=0.0,
    )
```

```python
result = docling_converter.convert(ppt_path) #! ERROR: ppt file not supported 
```

```python
result = docling_converter.convert(pptx_path)
result_docling = result.document.export_to_markdown()
```

```python
from rich.markdown import Markdown as RichMarkdown

md = RichMarkdown(result_docling)
display(md)
```

WITH IMAGE DESCRIPTION

```python
docling_converter = DoclingVLMConverter(
    model= "Qwen3-VL-8B-Instruct",
    hostname_and_port="https://llmservice.air.id/chat/completions",
    prompt = "Convert this page to markdown.",
    api_key="",
    response_format = "markdown",
    temperature=0.0,
    enable_picture_description=True,
    picture_prompt="Describe the image and flowchart in detail.",
    )
result = docling_converter.convert(pptx_path)
result_docling = result.document.export_to_markdown()
```

```python
from rich.markdown import Markdown as RichMarkdown

md = RichMarkdown(result_docling)
display(md)
```
