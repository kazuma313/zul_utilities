```python
from zul.utilities.OCR.docling_OCR import DoclingVLMConverter
```

```python
image_path = r"\tutorial\assets\ocr\table.jpeg"
image_jelek_path = r"\tutorial\assets\ocr\table_jelek.jpeg"
```

### granate-docling-258M

```python
from docling.datamodel.pipeline_options_vlm_model import ResponseFormat

docling_converter = DoclingVLMConverter(
    model= "granite-docling-258M",
    hostname_and_port="",
    prompt = "Convert this page to docling.",
    response_format = "doctags",
    temperature=0.0
    )
```

```python
result = docling_converter.convert(image_path)
result_docling = result.document.export_to_markdown()
```

```python
result_docling
```

```python
from rich.markdown import Markdown as RichMarkdown


md = RichMarkdown(result_docling)
display(md)
```

```python
print(result_docling.replace('\\n', '\n'))
```
