```python
from zul.utilities.OCR.docling_OCR import DoclingVLMConverter
from rich.markdown import Markdown as RichMarkdown
```

```python
file_path = r""
output_path = r""
```

### olmOCR-7B

```python
docling_converter = DoclingVLMConverter(
    model= "olmOCR-7B-0225-preview",
    hostname_and_port="",
    prompt = "Convert this page to MARKDOWN.",
    response_format = "markdown",
    temperature=0.0
    )
```

```python
result = docling_converter.convert(file_path)
```

inference: 9 minute 49.5 seconds

```python
result_docling = result.document.export_to_markdown()
```

```python
md = RichMarkdown(result_docling)
display(md)
```

```python
file_output_name =output_path.format(nama_file=model_name)
with open(file_output_name, "w", encoding="utf-8") as file:
    file.write(result_docling)
print(f"Markdown file {file_output_name}.md created successfully.")
```
