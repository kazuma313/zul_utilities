```python
import fitz  # PyMuPDF

def pdf_to_html(pdf_path, output_dir="output"):
    doc = fitz.open(pdf_path)
    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        html = page.get_text("html")  # Extracts layout-preserving HTML
        output_file = f"{output_dir}/page-{page_num + 1}.html"
        with open(output_file, "w") as f:
            f.write(html)
    doc.close()

# Usage
# pdf_to_html("input.pdf")
```
