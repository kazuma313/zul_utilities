```python
milvus_helper = MilvusHelper(config_path=r"boilerplate\tutorial\assets\milvus\config.json")
```

```python
from zul.utilities.fake_embedding import FakeEmbeddingModel
EMBEDDING = FakeEmbeddingModel(dimension=768, seed=42)
EMBEDDING.encode("hallo").flatten().shape
```

```python
data = [
    {"embedding": EMBEDDING.encode("hallo").flatten()},
    {"embedding": EMBEDDING.encode("hai").flatten()}
]
# data = {"id": "1", "embedding": EMBEDDING.encode("hallo").flatten()}

# data = {
#     "embedding": [0.1] * 768
# }

milvus_helper.insert(collection_name="my_collection", data=data)
```

```python
query_vector= EMBEDDING.encode("lah").flatten()
milvus_helper.search(collection_name="my_collection", 
                     query_vectors=[query_vector],
                     anns_field="embedding",
                     output_fields=["id"]
                     )
```
