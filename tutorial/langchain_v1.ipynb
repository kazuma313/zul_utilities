```python
from langchain.agents import create_agent
from langchain.agents.middleware import after_model, AgentState
from langgraph.runtime import Runtime
from typing import Any
```

```python
@after_model
def handle_empty_response(state: AgentState, runtime: Runtime) -> dict[str, Any] | None:
    """Handle empty string response dari model."""
    messages = state["messages"]
    last_message = messages[-1]
    if hasattr(last_message, 'content') and last_message.content == "":
        last_message.content = "panggil tool"
        return {"messages": [last_message]}
    return None
```

```python
from langchain_openai import ChatOpenAI

llm_model = ChatOpenAI(
    base_url="https://llmservice.air.id",
    api_key="",  # if your API doesn't require a real key, use a placeholder
    model="qwen2-32B",
    temperature=0.1,
) 

llm_model.invoke("What is the capital of France?")
```

```python
def get_weather(city: str) -> str:  
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"
```

```python
agent = create_agent(
    model=llm_model,
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
    middleware=[handle_empty_response]  # Pass middleware
)

# Run the agent
result = agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]}
)
```
